```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv

load_dotenv()
google_api_key = os.getenv('GOOGLE_API_KEY')
chat = ChatGoogleGenerativeAI(model = 'gemini-2.0-flash', temperature = 0)
```

```python
from pydantic import Field, BaseModel

class WeatherSearch(BaseModel):
    """Call this with an airport code to get the weather at the airport."""
    airport_code: str = Field(description="airport code for which we need to get the weather")

from langchain.utils.openai_functions import convert_pydantic_to_openai_function
weather_function = WeatherSearch.model_json_schema()
response = chat.invoke("weather in DEL?", functions = [weather_function])
```

```python
print(response)
```

```python
model_with_func = chat.bind(functions = [weather_function])

from langchain.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_template("You're helpful assistant. Answer the user queries: {queries}")

chain = prompt | model_with_func
response_1 = chain.invoke({'queries': "what's weather in DEL?"})
print(response_1)
```

```python
class ArtistSearch(BaseModel):
    """Call this to get the names of songs by a particular artist"""
    artist_name: str = Field(description="name of artist to look up")
    n: int = Field(description="number of results")

functions = [
    weather_function, ArtistSearch.model_json_schema()
]

model_with_funcs = chat.bind(functions = functions)

chain_1 = prompt | model_with_funcs
response_2 = chain_1.invoke({'queries': "what's weather in DEL?"})
response_3 = chain_1.invoke({'queries': "three songs by Taylor Swift"})
print(response_2)
print(response_3)
```
